```python
from agno.agent import Agent
from agno.models.openai import OpenAILike
from agno.tools.duckduckgo import DuckDuckGoTools
from agno.workflow import Workflow
from agno.db.sqlite import SqliteDb
from agno.knowledge.knowledge import Knowledge
from dotenv import load_dotenv
import os
load_dotenv("key.env")
api_key = os.getenv("OPENAI_API_KEY")
```

Flow search

```python
for k in [
    "HTTP_PROXY", "HTTPS_PROXY", "ALL_PROXY",
    "http_proxy", "https_proxy", "all_proxy", 
]:
    os.environ.pop(k, None) # Я так и не понял, что там за проблемы с прокси, но вот так вроде работает ;(
model = OpenAILike(
    id="ai-sage/GigaChat3-10B-A1.8B",
    name="Foundation Model",
    provider="cloud.ru",
    supports_native_structured_outputs=False,
    api_key=api_key,
    base_url="https://foundation-models.api.cloud.ru/v1"
)
poiskovik = Agent(
    model=model,
    tools=[DuckDuckGoTools()],
    instructions=(
        "Ты - агент, который ищет ответ на вопрос пользователя через браузер DuckDuckGo."
        "Используй исключительно инструменты DuckDuckGo для поиска информации."
        "Используй только русский язык в ответах."
    ),
    markdown=True,
)

workflow = Workflow(
    name="workflow_duckduckgo",
    steps=[poiskovik]
)

result = workflow.print_response("Какое настоящее имя инстасамки?", stream=True)

```

Agent

```python
db = SqliteDb(db_file="agno.db")
knowledge = Knowledge()
knowledge.add_content(
    text_content= "Инстаса́мка (Instasamka; настоящее имя — Да́рья Евге́ньевна Еро́пкина[К 1]; род. 11 мая 2000, Тобольск, Россия) — российская поп-певица, рэп-исполнитель и тиктокер; ранее — инстаблогер. "
)
poiskovik = Agent(
    model=model,
    tools=[DuckDuckGoTools()],
    knowledge=knowledge,
    db=db,
    enable_user_memories=True,
    search_knowledge=True,
    instructions=(
        "Ты - агент, который ищет ответ на вопрос пользователя через браузер DuckDuckGo."
        "Перед тем, как ответить, проверяй базу знаний"
        "Включи в итоговый ответ как минимум один факт из базы знаний"
        "Используй только базу знаний и инструменты DuckDuckGo для поиска информации."
        "Используй только русский язык в ответах."
    ),
    markdown=True,
)
workflow = Workflow(
    name="workflow_duckduckgo",
    steps=[poiskovik]
)
result = workflow.print_response("расскажи о Инстасамке", stream=True)
```

MultiAgent

```python
knowledge = Knowledge()


knowledge.add_content(
    text_content="""
    Найди мне информацию о самых популярных событиях, связанных с Инстасамкой. Используй любые источники информации,
    видео на YouTube, телеграмм-каналы, новостные источники.
    """
)


solnik_finder = Agent(
    name="Solnik-finder",
    model=model,
    tools=[DuckDuckGoTools()],
    knowledge=knowledge,
    enable_user_memories=True,
    search_knowledge=True,
    instructions=(
        "Ты ищешь музыкальные релизы, клипы и крупные инфоповоды, связанные с Инстасамкой, на всех доступных платформах.",
        "Всегда фокусируйся только на певице Инстасамке (Instasamka).",
        "Возвращай результаты в формате: Тип события (трек/альбом/клип/новость), название, год, площадка/источник, основные темы или инфоповод, примерная статистика просмотров/прослушиваний, если есть."
    ),
    markdown=True, 
)


info_finder = Agent(
    name="Info-finder",
    model=model,
    tools=[DuckDuckGoTools()],
    knowledge=knowledge,
    enable_user_memories=True,
    search_knowledge=True,
    instructions=(
        "Ты ищешь биографическую и контекстную информацию об Инстасамке (Instasamka).",
        "Собирай данные о её настоящем имени, возрасте, дате и месте рождения, текущей стране проживания, основных этапах карьеры, музыкальном стиле и репутации в медиа.",
        "Возвращай результаты в формате: параметр — значение, включая: настоящее имя, дата рождения, город рождения, текущая страна проживания, ключевые релизы, известные скандалы или инфоповоды (кратко)."
    ),
    markdown=True, 
)


workflow = Workflow(
    name="instasamka-info",
    steps=[solnik_finder,
           info_finder]
)


result = workflow.print_response("Расскажи о том, кто такая инстасамка и какие у неё были крупные события в карьере", stream=True)
```
